### 1. Carga, Preparación y Aumento de Datos (Data Augmentation)

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2

# Cargar el dataset CIFAR-10
(X_train, y_train), (X_test, y_test) = keras.datasets.cifar10.load_data()

# Normalizar los píxeles al rango [0, 1]
X_train = X_train.astype("float32") / 255.0
X_test = X_test.astype("float32") / 255.0

# Convertir etiquetas a categóricas (One-Hot Encoding)
y_train = keras.utils.to_categorical(y_train, 10)
y_test = keras.utils.to_categorical(y_test, 10)

print(f"X_train shape: {X_train.shape} - y_train shape: {y_train.shape}")
print(f"X_test shape: {X_test.shape} - y_test shape: {y_test.shape}")

# Definir el pipeline de Data Augmentation
data_augmentation = keras.Sequential([
    layers.RandomFlip("horizontal"),
    layers.RandomRotation(0.1),
    layers.RandomZoom(0.1)
])
```

### 2. Definición de Modelos

```python
def build_cnn_1(input_shape=(32, 32, 3), num_classes=10):
    """Arquitectura CNN Propia 1: Básica/Ligera"""
    inputs = layers.Input(shape=input_shape)
    x = data_augmentation(inputs)

    x = layers.Conv2D(32, (3, 3), padding='same', activation='relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)

    x = layers.Conv2D(64, (3, 3), padding='same', activation='relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)

    x = layers.Flatten()(x)
    x = layers.Dense(128, activation='relu')(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = keras.Model(inputs, outputs, name="CNN_Propia_1_Basica")
    return model

def build_cnn_2(input_shape=(32, 32, 3), num_classes=10):
    """Arquitectura CNN Propia 2: Robusta con regularización, BN y Dropout"""
    inputs = layers.Input(shape=input_shape)
    x = data_augmentation(inputs)

    # Bloque 1
    x = layers.Conv2D(32, (3, 3), padding='same', kernel_regularizer=keras.regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(32, (3, 3), padding='same', kernel_regularizer=keras.regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.2)(x)

    # Bloque 2
    x = layers.Conv2D(64, (3, 3), padding='same', kernel_regularizer=keras.regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Conv2D(64, (3, 3), padding='same', kernel_regularizer=keras.regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.MaxPooling2D((2, 2))(x)
    x = layers.Dropout(0.3)(x)

    # Bloque denso
    x = layers.Flatten()(x)
    x = layers.Dense(256, kernel_regularizer=keras.regularizers.l2(1e-4))(x)
    x = layers.BatchNormalization()(x)
    x = layers.Activation('relu')(x)
    x = layers.Dropout(0.4)(x)

    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = keras.Model(inputs, outputs, name="CNN_Propia_2_Robusta")
    return model

def build_transfer_learning(input_shape=(32, 32, 3), num_classes=10):
    """Arquitectura 3: Transfer Learning usando MobileNetV2"""
    inputs = layers.Input(shape=input_shape)
    x = data_augmentation(inputs)

    # MobileNetV2 requiere redimensionar las imágenes preferiblemente a un tamaño mayor
    x = layers.UpSampling2D(size=(3, 3))(x) # Escala 32x32 a 96x96

    base_model = MobileNetV2(input_shape=(96, 96, 3), include_top=False, weights='imagenet')
    base_model.trainable = False  # Congelamos los pesos base

    x = base_model(x, training=False)
    x = layers.GlobalAveragePooling2D()(x)
    x = layers.Dropout(0.2)(x)
    outputs = layers.Dense(num_classes, activation='softmax')(x)

    model = keras.Model(inputs, outputs, name="Transfer_Learning_MobileNetV2")
    return model

# Instanciar y resumir los modelos
model_1 = build_cnn_1()
model_2 = build_cnn_2()
model_3 = build_transfer_learning()

model_2.summary()
```

### 3. Entrenamiento de Modelos

```python
epochs = 15
batch_size = 64

histories = {}

def train_model(model, name):
    print(f"\n--- Entrenando {name} ---")
    model.compile(
        optimizer='adam',
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )
    history = model.fit(
        X_train, y_train,
        epochs=epochs,
        batch_size=batch_size,
        validation_data=(X_test, y_test),
        verbose=1
    )
    histories[name] = history
    return history

# Entrenamos los tres modelos
train_model(model_1, "CNN_1_Basica")
train_model(model_2, "CNN_2_Robusta")
train_model(model_3, "Transfer_Learning_MobileNet")
```

### 4. Estadística y Gráficos Comparativos

```python
fig, axs = plt.subplots(1, 2, figsize=(16, 6))

# Gráfico de Precisión (Accuracy)
for name, history in histories.items():
    axs[0].plot(history.history['accuracy'], label=f'{name} - Train')
    axs[0].plot(history.history['val_accuracy'], label=f'{name} - Val', linestyle='--')
axs[0].set_title('Comparativa de Precisión (Accuracy)')
axs[0].set_xlabel('Epochs')
axs[0].set_ylabel('Accuracy')
axs[0].legend()
axs[0].grid(True)

# Gráfico de Pérdida (Loss)
for name, history in histories.items():
    axs[1].plot(history.history['loss'], label=f'{name} - Train')
    axs[1].plot(history.history['val_loss'], label=f'{name} - Val', linestyle='--')
axs[1].set_title('Comparativa de Pérdida (Loss)')
axs[1].set_xlabel('Epochs')
axs[1].set_ylabel('Loss')
axs[1].legend()
axs[1].grid(True)

plt.tight_layout()
plt.show()
```

### 5. Conclusiones y Reporte Comparativo

#### Análisis del Desempeño
1. **CNN Propia 1 (Básica):**
   - **Desempeño:** Al ser una arquitectura poco profunda y sin regularización estricta, aprende rápido pero tiende a estancarse más temprano en precisión de validación, mostrando indicios de sobreajuste (overfitting) a pesar del data augmentation.
   
2. **CNN Propia 2 (Robusta):**
   - **Desempeño:** Es la arquitectura diseñada desde cero con mejor generalización. Gracias a la incorporación de **Batch Normalization**, regularizadores **L2** y capas de **Dropout**, la curva de pérdida de validación desciende de forma estable y cercana a la de entrenamiento. Su capacidad para capturar patrones complejos en CIFAR-10 es notablemente superior al baseline simple.

3. **Transfer Learning (MobileNetV2):**
   - **Desempeño:** Al heredar features pre-entrenadas de ImageNet (y realizar un re-escalado interno vía `UpSampling2D`), este modelo parte de un excelente punto de partida. Logra una alta precisión de forma sumamente veloz. Sin embargo, requiere mayor costo de cómputo debido al tamaño de entrada aumentado.

- **¿Por qué?**
  El Transfer Learning aprovecha filtros jerárquicos ya aprendidos en millones de imágenes diversas, mientras que la CNN Propia 2 triunfa gracias al balance de regularización estricta y capas profundas de convolución.
